```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# --------------------------------
# Step 1: Input Dataset
# --------------------------------

reviews = [
    "The product quality is excellent",
    "I am very happy with the service",
    "The delivery was very slow",
    "The product is damaged",
    "Excellent customer support",
    "Very disappointed with the purchase"
]

# Positive = 1, Negative = 0
labels = np.array([1, 1, 0, 0, 1, 0])

# --------------------------------
# Step 2: Tokenization
# --------------------------------

tokenizer = Tokenizer()
tokenizer.fit_on_texts(reviews)

sequences = tokenizer.texts_to_sequences(reviews)

print("Tokenized Sequences:")
print(sequences)

# --------------------------------
# Step 3: Padding
# --------------------------------

max_length = max(len(seq) for seq in sequences)

X = pad_sequences(
    sequences,
    maxlen=max_length,
    padding='post'
)

print("\nPadded Sequences:")
print(X)

# --------------------------------
# Step 4: Build LSTM Model
# --------------------------------

vocab_size = len(tokenizer.word_index) + 1

model = Sequential([
    Embedding(
        input_dim=vocab_size,
        output_dim=16,
        input_length=max_length
    ),
    LSTM(32),
    Dense(1, activation='sigmoid')
])

# --------------------------------
# Step 5: Compile Model
# --------------------------------

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# --------------------------------
# Step 6: Train Model
# --------------------------------

model.fit(
    X,
    labels,
    epochs=100,
    verbose=0
)

print("\nLSTM model trained successfully.")

# --------------------------------
# Step 7: Predict New Review
# --------------------------------

new_review = ["The product is very good and easy to use"]

new_sequence = tokenizer.texts_to_sequences(new_review)

new_padded = pad_sequences(
    new_sequence,
    maxlen=max_length,
    padding='post'
)

prediction = model.predict(new_padded, verbose=0)[0][0]

print("\nNew Review:")
print(new_review[0])

print("Prediction Probability:", round(float(prediction), 4))

if prediction >= 0.5:
    print("Predicted Sentiment: Positive")
else:
    print("Predicted Sentiment: Negative")
```
