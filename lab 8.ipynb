```python
from gensim.models import Word2Vec
import re

# -------------------------------
# Activity 1: Data Preparation
# -------------------------------

corpus = [
    "Machine learning is a branch of artificial intelligence.",
    "Artificial intelligence helps computers learn from data.",
    "Deep learning is a subset of machine learning.",
    "Natural language processing uses machine learning."
]

# Tokenization
tokenized_corpus = [
    re.findall(r'\b\w+\b', sentence.lower())
    for sentence in corpus
]

print("Tokenized Corpus:")
for sentence in tokenized_corpus:
    print(sentence)

# -------------------------------
# Activity 2: Train CBOW Model
# -------------------------------

model = Word2Vec(
    sentences=tokenized_corpus,
    vector_size=100,
    window=2,
    min_count=1,
    workers=1,
    sg=0,
    seed=42,
    epochs=100
)

print("\nVocabulary:")
print(list(model.wv.index_to_key))

print("\nWord2Vec CBOW model trained successfully!")

# -------------------------------
# Activity 3: Word Embeddings
# -------------------------------

machine_vector = model.wv['machine']
learning_vector = model.wv['learning']

print("\nEmbedding vector for 'Machine':")
print(machine_vector)

print("\nEmbedding vector for 'Learning':")
print(learning_vector)

print("\nVector dimension:", len(machine_vector))

# -------------------------------
# Activity 4: Similarity Analysis
# -------------------------------

print("\nTop 5 words similar to 'Learning':")

for word, score in model.wv.most_similar('learning', topn=5):
    print(word, ":", round(score, 4))

similarity1 = model.wv.similarity('machine', 'learning')
similarity2 = model.wv.similarity('artificial', 'intelligence')

print("\nSimilarity between Machine and Learning:",
      round(similarity1, 4))

print("Similarity between Artificial and Intelligence:",
      round(similarity2, 4))
```
