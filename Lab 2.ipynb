```python
import nltk
import string
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')
```

```python
text = "Dr.Ravi is studying Machine Learning at Anna University in Chennai."
```

```python
tokens = word_tokenize(text)
```

```python
stop_words = set(stopwords.words('english'))
filtered_tokens = [
    word
    for word in tokens
    if word.lower() not in stop_words
    and word not in string.punctuation
]
```

```python
print("orginal tokens:")
print(tokens)
print("\nTokens after stop-word & punctuation Removal:")
print(filtered_tokens)
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import string
```

```python
text = "Dr.Ravi is studying Machine Learning at Anna University in Chennai."
```

```python
tokens = word_tokenize(text)
```

```python
stop_words = set(stopwords.words('english'))
filtered_tokens = [
    word
    for word in tokens
    if word.lower() not in stop_words
    and word not in string.punctuation
]
```

```python
stemmer = PorterStemmer()
stemmed_tokens = [
    stemmer.stem(word.lower())
    for word in filtered_tokens
]
```

```python
print("Tokens after stop-word & punctuation Removal:")
print(filtered_tokens)
print("\nStemmed tokens")
print(stemmed_tokens)
```
