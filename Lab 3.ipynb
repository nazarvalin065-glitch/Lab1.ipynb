```python
import numpy as np
import pandas as pd

# Given corpus
corpus = [
    "I love NLP",
    "I love Python",
    "NLP is fun"
]

# Create vocabulary
words = set(" ".join(corpus).lower().split())
vocabulary = sorted(words)

print("Vocabulary:", vocabulary)

# One-Hot Encoding
data = []
for word in vocabulary:
    vector = [1 if word == w else 0 for w in vocabulary]
    data.append(vector)

df = pd.DataFrame(data, index=vocabulary, columns=vocabulary)

print("\nOne-Hot Encoding:")
print(df)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Create BoW model
vectorizer = CountVectorizer()

# Create DTM
bow_matrix = vectorizer.fit_transform(corpus)

# Vocabulary
print("Vocabulary:")
print(vectorizer.get_feature_names_out())

# BoW Matrix
print("\nBag of Words Matrix:")
print(bow_matrix.toarray())
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from collections import Counter

# Combine all documents
text = " ".join(corpus)

# Generate Word Cloud
wordcloud = WordCloud(width=800, height=400,
                      background_color="white").generate(text)

# Display Word Cloud
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()

# Find top 5 frequent words
words = text.lower().split()
frequency = Counter(words)

print("Top 5 frequent words:")
print(frequency.most_common(5))
```
