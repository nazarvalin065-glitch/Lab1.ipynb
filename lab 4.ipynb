```python
import pandas as pd

# Create input corpus
documents = [
    "Apple is sweet",
    "Apple is red",
    "Mango is sweet"
]

# Generate vocabulary
vocabulary = set()

for document in documents:
    words = document.lower().split()
    vocabulary.update(words)

# Convert set to sorted list
vocabulary = sorted(vocabulary)

# Display vocabulary
print("Vocabulary:")
print(vocabulary)

print("Vocabulary Size:", len(vocabulary))
```

```python
# Import required libraries
import pandas as pd
import math

# Input corpus
documents = [
    "Apple is sweet",
    "Apple is red",
    "Mango is sweet"
]

# Create vocabulary
vocabulary = sorted(set(
    word.lower()
    for document in documents
    for word in document.split()
))

# --------------------------------
# 1. Total words in each document
# --------------------------------

total_words = []

for document in documents:
    words = document.lower().split()
    total_words.append(len(words))

print("Total words in each document:")
print(total_words)


# --------------------------------
# 2. Calculate Term Frequency (TF)
# --------------------------------

tf_matrix = []

for document in documents:
    words = document.lower().split()
    total = len(words)

    tf_values = []

    for term in vocabulary:
        count = words.count(term)
        tf = count / total
        tf_values.append(tf)

    tf_matrix.append(tf_values)


# --------------------------------
# 3. Display TF Matrix
# --------------------------------

tf_df = pd.DataFrame(
    tf_matrix,
    columns=vocabulary,
    index=["D1", "D2", "D3"]
)

print("\nTF Matrix:")
print(tf_df.round(3))


# --------------------------------
# 4. Calculate Document Frequency
# --------------------------------

df_values = []

for term in vocabulary:
    count = 0

    for document in documents:
        words = document.lower().split()

        if term in words:
            count += 1

    df_values.append(count)


# --------------------------------
# 5. Calculate IDF
# --------------------------------

N = len(documents)

idf_values = []

for df in df_values:
    idf = math.log10(N / df)
    idf_values.append(idf)


# --------------------------------
# 6. Display DF and IDF
# --------------------------------

df_idf = pd.DataFrame({
    "Term": vocabulary,
    "DF": df_values,
    "IDF": idf_values
})

print("\nDF and IDF:")
print(df_idf.round(3))


# --------------------------------
# TF-IDF Matrix
# --------------------------------

tfidf_matrix = []

for i in range(len(documents)):
    row = []

    for j in range(len(vocabulary)):
        tfidf = tf_matrix[i][j] * idf_values[j]
        row.append(tfidf)

    tfidf_matrix.append(row)

tfidf_df = pd.DataFrame(
    tfidf_matrix,
    columns=vocabulary,
    index=["D1", "D2", "D3"]
)

print("\nTF-IDF Matrix:")
print(tfidf_df.round(3))
```
